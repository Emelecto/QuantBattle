# C6-L6 · Bitácora de trades
30 trades demo con plan, emoción y resultado en R: la disciplina se mide, no se declara.

```python
import pandas as pd, numpy as np
from pathlib import Path
URL = 'https://raw.githubusercontent.com/Emelecto/QuantBattle/main/web/content/cursos/torneos/data/c6_l6.csv'
try:
    df = pd.read_csv(URL)
    print('Fuente: URL (Colab)')
except Exception as e:
    print('Sin red, uso fallback local:', e)
    for cand in [Path('../data/c6_l6.csv'), Path('data/c6_l6.csv'), Path('c6_l6.csv')]:
        if cand.exists():
            df = pd.read_csv(cand); break
    print('Fuente: local')
print(df.shape)
print(df.head(5).to_string(index=False))
```

```python
plan = df[df.siguio_plan == 1]['resultado_R']
fuera = df[df.siguio_plan == 0]['resultado_R']
print('con plan %+.2fR (n=%d) · sin plan %+.2fR (n=%d)' % (plan.mean(), len(plan), fuera.mean(), len(fuera)))
print(df.groupby('setup')['resultado_R'].agg(['mean', 'count']).round(2).to_string())
```

```python
imp = df[df.siguio_plan == 0].groupby('emocion').size()
print('impulsivas por emocion:')
print(imp.to_string())
print('winrate con plan: %.0f%% · sin plan: %.0f%%' % (float((plan > 0).mean() * 100), float((fuera > 0).mean() * 100)))
```

```python
assert len(df) == 30 and set(['setup', 'resultado_R', 'emocion', 'siguio_plan']) <= set(df.columns)
assert abs(plan.mean() - 0.41) < 0.05 and abs(fuera.mean() + 0.22) < 0.05
assert df['resultado_R'].between(-2.0, 2.5).all()
print('OK L6: bitacora verificada, con plan %+.2fR vs sin plan %+.2fR' % (plan.mean(), fuera.mean()))
```
