# C2-L1 · Tu laboratorio: venv, Jupyter y pandas
Carga un CSV de precios, inspecciónalo y grafícalo. Funciona en Colab y en local.

```python
import pandas as pd
from pathlib import Path

URL = 'https://raw.githubusercontent.com/Emelecto/QuantBattle/main/web/content/cursos/python/data/c2_l1.csv'
csv_path = Path('c2_l1.csv')
try:
    df = pd.read_csv(URL, parse_dates=['date'])
    print('Fuente: URL (Colab)')
except Exception as e:
    print('Sin red, uso fallback local:', e)
    for cand in [Path('../data/c2_l1.csv'), Path('data/c2_l1.csv'), Path('c2_l1.csv')]:
        if cand.exists():
            csv_path = cand; break
    df = pd.read_csv(csv_path, parse_dates=['date'])
    print('Fuente: local', csv_path)
print(df.shape)
```

```python
print(df.head(3).to_string(index=False))
print()
print(df[['close','volume']].describe().round(2).to_string())
print('\nTipos:')
print(df.dtypes)
```

## Gráfico: precio y volumen

```python
import matplotlib.pyplot as plt
df.plot(x='date', y='close', title='C2-L1 · Cierre diario')
plt.show()
print(f"días={len(df)}  cierre_ini={df['close'].iloc[0]:.2f}  cierre_fin={df['close'].iloc[-1]:.2f}")
```
