# C3-L2 · Media vs mediana
Estadística descriptiva robusta: 50 retornos con 2 días extremos.

```python
import pandas as pd, numpy as np
from pathlib import Path

URL = 'https://raw.githubusercontent.com/Emelecto/QuantBattle/main/web/content/cursos/mates/data/c3_l2.csv'
try:
    df = pd.read_csv(URL)
    print('Fuente: URL (Colab)')
except Exception as e:
    print('Sin red, uso fallback local:', e)
    for cand in [Path('../data/c3_l2.csv'), Path('data/c3_l2.csv'), Path('c3_l2.csv')]:
        if cand.exists():
            df = pd.read_csv(cand); break
    print('Fuente: local')
print(df.shape)
```

```python
r = df['ret'].to_numpy()
print(f'media={r.mean():.4f}  mediana={np.median(r):.4f}')
print(f'std={r.std(ddof=1):.4f}')
p5,p25,p50,p75,p95 = np.percentile(r,[5,25,50,75,95])
print(f'P5={p5:.4f} P25={p25:.4f} P50={p50:.4f} P75={p75:.4f} P95={p95:.4f}')
print(f'IQR={p75-p25:.4f}')
```

```python
r2 = df['ret'].sort_values().iloc[1:-1].to_numpy()  # sin los 2 extremos
print(f'sin extremos: media={r2.mean():.4f} mediana={np.median(r2):.4f}')
print(f'la media se mueve {(r.mean()-r2.mean()):.4f}; la mediana {(np.median(r)-np.median(r2)):.4f}')
```

```python
# Chequeo automático
assert len(df) == 50
assert abs(np.median(r) - np.percentile(r, 50)) < 1e-12
assert (np.percentile(r,75)-np.percentile(r,25)) > 0
print('OK: descriptiva verificada')
```
