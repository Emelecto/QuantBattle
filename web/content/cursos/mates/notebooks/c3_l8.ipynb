# C3-L8 · Inferencia y p-valor | Test t de media sobre 50 retornos diarios.

```python
# CELDA COLAB-FIRST: correla primero si estas en Google Colab.
# Descarga el CSV del repo; si falla (sin red ), usa el CSV local.
import pandas as pd
from pathlib import Path

ORG = "Emelecto"  # organizacion fija del repo Emelecto/QuantBattle
CSV_NOMBRE = "c3_l8.csv"
CSV_URL = f"https://raw.githubusercontent.com/{ORG}/QuantBattle/main/web/content/cursos/mates/data/{CSV_NOMBRE}"

try:
    df = pd.read_csv(CSV_URL)
    print("CSV descargado desde:", CSV_URL)
except Exception as e:
    print("Uso CSV local (motivo:", str(e)[:80] + ")")
    csv_path = Path("../data") / CSV_NOMBRE
    if not csv_path.exists():
        csv_path = Path(CSV_NOMBRE)  # fallback si corres desde data/
    df = pd.read_csv(csv_path)
print(df.shape)
print(df.head())
```

```python
import numpy as np, math
r = df["ret"]
n = len(r)
media = r.mean()
se = r.std(ddof=1) / math.sqrt(n)
t = media / se
Phi = lambda x: 0.5 * (1 + math.erf(x / math.sqrt(2)))
p = 2 * (1 - Phi(abs(t)))
print(f"Media diaria: {media:.4f} | n = {n}")
print(f"Estadístico t: {t:.2f}")
print(f"p-valor bilateral: {p:.3f}")
print("Rechaza suerte al 5%:" , p < 0.05)
```

## ¿Suerte o real? | t = 3,15 cae lejos en la cola: p = 0,002.

```python
import matplotlib.pyplot as plt
xs = np.linspace(-4, 4, 400)
ys = np.exp(-xs ** 2 / 2) / np.sqrt(2 * np.pi)
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(xs, ys, color="#52525b")
ax.axvline(t, color="#f87171", linewidth=2, label=f"t = {t:.2f}")
ax.axvline(0, color="#5eead4", linestyle="--", label="nula: media 0")
ax.set_title("t = 3,15 en la cola: p = 0,002, suerte descartada")
ax.legend()
plt.show()
```

```python
print("p = 0,002: ver esto por azar pasaría 2 de cada 1000 veces.")
print("Significativo ≠ rentable (faltan costos) ni eterno (el régimen cambia).")
```
