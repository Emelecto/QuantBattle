# C3-L3 · Log-retornos con BNB
Simples vs log con 50 cierres de BNB. La suma de logs clava el total.

```python
import pandas as pd, numpy as np
from pathlib import Path

URL = 'https://raw.githubusercontent.com/Emelecto/QuantBattle/main/web/content/cursos/mates/data/c3_l3.csv'
try:
    df = pd.read_csv(URL)
    print('Fuente: URL (Colab)')
except Exception as e:
    print('Sin red, uso fallback local:', e)
    for cand in [Path('../data/c3_l3.csv'), Path('data/c3_l3.csv'), Path('c3_l3.csv')]:
        if cand.exists():
            df = pd.read_csv(cand); break
    print('Fuente: local')
print(df.shape)
```

```python
df['ret_simple'] = df['close'].pct_change()
df['ret_log'] = np.log(df['close'] / df['close'].shift(1))
print(df[['dia','close','ret_simple','ret_log']].head(8).to_string(index=False))
print(f"dif media abs(simple-log)={(df['ret_simple']-df['ret_log']).abs().mean():.6f}")
```

```python
suma_logs = df['ret_log'].sum()
total = np.log(df['close'].iloc[-1] / df['close'].iloc[0])
comp = (1 + df['ret_simple'].dropna()).prod() - 1
print(f'suma_logs={suma_logs:.6f}  ln(Pn/P0)={total:.6f}  dif={abs(suma_logs-total):.2e}')
print(f'compuesto simples={comp:.4f}  exp(suma_logs)-1={np.exp(suma_logs)-1:.4f}')
```

```python
# Chequeo automático
assert abs(df['ret_log'].sum() - np.log(df['close'].iloc[-1]/df['close'].iloc[0])) < 1e-9, 'los logs deben sumar el total'
assert abs((1+0.10)*(1-0.10) - 0.99) < 1e-12
print('OK: log-retornos verificados')
```
