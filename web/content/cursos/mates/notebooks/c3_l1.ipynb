# C3-L1 · Probabilidad, Bayes y valor esperado
Win rate, Bayes y EV con 50 trades. Incluimos costos antes de decidir.

```python
import pandas as pd, numpy as np
from pathlib import Path

URL = 'https://raw.githubusercontent.com/Emelecto/QuantBattle/main/web/content/cursos/mates/data/c3_l1.csv'
try:
    df = pd.read_csv(URL)
    print('Fuente: URL (Colab)')
except Exception as e:
    print('Sin red, uso fallback local:', e)
    for cand in [Path('../data/c3_l1.csv'), Path('data/c3_l1.csv'), Path('c3_l1.csv')]:
        if cand.exists():
            df = pd.read_csv(cand); break
    print('Fuente: local')
print(df.shape)
```

```python
p = df['win'].mean()
gan = df.loc[df['win']==1,'pnl'].mean()
per = df.loc[df['win']==0,'pnl'].mean()
ev = p*gan + (1-p)*per
print(f'win_rate={p:.2f}  gan_media={gan:.1f}  per_media={per:.1f}')
print(f'EV_bruto={ev:.2f} USD/trade')
```

```python
prior = df['win'].mean()
a = (df['vol_confirm']==1).mean()
b = df.loc[df['win']==1,'vol_confirm'].mean()
c = df.loc[df['win']==0,'vol_confirm'].mean()
post = b*prior/((b*prior)+(c*(1-prior)))
print(f'P(confirm)={a:.2f}  P(conf|gana)={b:.2f}  P(conf|pierde)={c:.2f}')
print(f'prior={prior:.2f} -> posterior={post:.2f}')
```

```python
for costo in [2.0, 5.0]:
    print(f'costo={costo}: EV_neto={ev-costo:.2f}  operable={(ev-costo)>0}')
```

```python
# Chequeo automático
assert len(df) == 50, '50 trades'
assert abs(df['win'].mean() - (df['pnl']>0).mean()) < 1e-9
p2 = df['win'].mean()
assert abs((p2*df.loc[df['win']==1,'pnl'].mean() + (1-p2)*df.loc[df['win']==0,'pnl'].mean()) - df['pnl'].mean()) < 1e-9
print('OK: EV verificado')
```
